```python
def binary_search(arr,target):
   low=0
   high=len(arr)-1
   while low<=high:
         mid=(low+high)//2
         if arr[mid]==target:
               return mid
         elif arr[mid]< target:
                low=mid+1
         else:
                high=mid-1
   return-1
sorted_data=[12,15,17,21,26,29,30]
print('Index of 21:',binary_search(sorted_data,21))
print('Index of 30:',binary_search(sorted_data,30))
```
