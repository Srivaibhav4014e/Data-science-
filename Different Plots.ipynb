```python

import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Four Visualization Techniques', fontsize=16, fontweight='bold')
np.random.seed(42)

# ── 1. Scatter Plot ──────────────────────────────
x = np.random.randn(100)
y = 2*x + np.random.randn(100)*0.5
axes[0,0].scatter(x, y, c='steelblue', alpha=0.7, s=50, edgecolors='white')
axes[0,0].set_title('Scatter Plot (Correlation r≈0.97)')
axes[0,0].set_xlabel('X'); axes[0,0].set_ylabel('Y')
# Add trend line
m,b = np.polyfit(x, y, 1)
axes[0,0].plot(np.sort(x), m*np.sort(x)+b, 'r-', linewidth=2, label='Trend')
axes[0,0].legend()

# ── 2. Histogram ─────────────────────────────────
scores = np.random.normal(70, 10, 500)
axes[0,1].hist(scores, bins=20, color='green',
               edgecolor='white', alpha=0.7, density=True)
axes[0,1].axvline(scores.mean(), color='red',
                  linestyle='--', label=f'Mean={scores.mean():.1f}')
axes[0,1].set_title('Histogram — Normal Distribution')
axes[0,1].set_xlabel('Score'); axes[0,1].set_ylabel('Density')
axes[0,1].legend()

# ── 3. Pie Chart ─────────────────────────────────
labels  = ['Python','Java','C++','JavaScript','Others']
sizes   = [35, 25, 20, 15, 5]
explode = (0.1, 0, 0, 0, 0)    # Explode Python slice
axes[1,0].pie(sizes, explode=explode, labels=labels,
              autopct='%1.1f%%', startangle=90,
              colors=['#FF6B6B','#4ECDC4','#45B7D1','#96CEB4','#FFEAA7'])
axes[1,0].set_title('Programming Language Popularity')

# ── 4. Box Plot ──────────────────────────────────
group_a = np.random.normal(65, 10, 50)
group_b = np.random.normal(75, 8, 50)
group_c = np.random.normal(80, 12, 50)
bp = axes[1,1].boxplot([group_a,group_b,group_c],
                        labels=['Class A','Class B','Class C'],
                        patch_artist=True)
colors = ['#FF6B6B','#4ECDC4','#45B7D1']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
axes[1,1].set_title('Score Distribution (Box Plot)')
axes[1,1].set_ylabel('Scores')
axes[1,1].grid(axis='y', alpha=0.3)

plt.tight_layout(); plt.show()
```
