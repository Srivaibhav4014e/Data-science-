```python
def bubble_sort(arr):
   n=len(arr)
   for i in range(n-1):
    swapped=False
    for j in range(n-1-i):
     if arr[j]>arr[j+1]:
       arr[j],arr[j+1]=arr[j+1],arr[j]
       swapped=True
    if swapped==False:
        break
   return arr
Data=[10,4,56,3,5,21]
print('sorted array:',bubble_sort(Data))
```
