```python
import numpy as np

from scipy import stats



np.random.seed(42)



# Generate random samples

normal_data  = np.random.normal(loc=50, scale=10, size=1000)

uniform_data = np.random.uniform(low=0, high=100, size=1000)

poisson_data = np.random.poisson(lam=5, size=1000)

exp_data     = np.random.exponential(scale=2, size=1000)

binom_data   = np.random.binomial(n=20, p=0.5, size=1000)



# Test if data follows a distribution

stat, p = stats.normaltest(normal_data)

print(f'Normality test: stat={stat:.4f}, p={p:.4f}')

print('Normal' if p > 0.05 else 'Not Normal')



stat, p = stats.kstest(uniform_data, 'uniform',

                        args=(0, 100))

print(f'KS test (uniform): stat={stat:.4f}, p={p:.4f}')
```
